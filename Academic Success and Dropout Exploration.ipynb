# Academic Success and Dropout Exploration and Data Analysis

## Background

In this project, we are exploring a dataset created from a higher education institution (acquired from several disjoint databases) related to students enrolled in different undergraduate degrees, such as agronomy, design, education, nursing, journalism, management, social service, and technologies. It includes information known at the time of student enrollment (academic path, demographics, and social-economic factors) and the students' academic performance at the end of the first and second semesters. The data is used to build classification models to predict students' dropout and academic sucess. The problem is formulated as a three category classification task, in which there is a strong imbalance towards one of the classes.

```python
# Can we identify at-risk students at the point of admission, or does early in-course performance tell a different story?

color_map = {'Dropout': '#ff6584', 'Enrolled': '#43e8d8', 'Graduate': '#6c63ff'}

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

for ax, sem in zip(axes, ['1st', '2nd']):
    approved_col = f'Curricular units {sem} sem (approved)'

    for status, color in color_map.items():
        subset = df[df['Target'] == status]
        ax.scatter(
            subset['Admission grade'],
            subset[approved_col],
            c=color, label=status, alpha=0.35, s=15, edgecolors='none'
        )

    ax.set_title(f'Admission Grade vs {sem} Sem Units Passed', fontsize=13, fontweight='bold', pad=12)
    ax.set_xlabel('Admission Grade', fontsize=11)
    ax.set_ylabel(f'Units Passed ({sem} Sem)', fontsize=11)
    ax.spines[['top', 'right']].set_visible(False)
    ax.legend(fontsize=9, markerscale=2)

plt.suptitle('Admission Grade vs Units Passed by Student Outcome', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

The scatter plots above reveal a critical insight for early dropout intervention: admission grade alone is a weak predictor of student success. Dropout students (pink) are spread across the full range of admission grades (95–190), meaning a student can enter the program with a competitive score and still pass little to nothing in their first semester. What clearly separates outcomes is early in-course performance — graduates consistently pass more units in both semesters, while dropouts cluster heavily at or near zero units passed. The horizontal banding in the plots reflects the discrete nature of unit completion (students pass whole courses, not fractions), further highlighting how starkly dropout students underperform from the start. This suggests that the most effective intervention window is not at the point of admission, but rather early in the first semester, where low unit completion rates can serve as an early warning signal for students who may need additional support.

```python
# Honing in Specifically on Dropouts: At what point did dropout students start disengaging? Let's look at the amount of units passed by students in the first and second semesters compared to how many credit units they were enrolled in.

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

dropouts = df[df['Target'] == 'Dropout'].copy()
graduates = df[df['Target'] == 'Graduate'].copy()

fig, axes = plt.subplots(2, 2, figsize=(14, 12))

for col, (group, label, cmap) in enumerate(zip(
    [dropouts, graduates],
    ['Dropouts', 'Graduates'],
    ['Reds', 'Reds']
)):
    for row, sem in enumerate(['1st', '2nd']):
        ax = axes[row][col]
        enrolled_col = f'Curricular units {sem} sem (enrolled)'
        approved_col = f'Curricular units {sem} sem (approved)'

        filtered = group[
            (group[enrolled_col] <= 9) &
            (group[approved_col] <= 9)
        ]

        heatmap_data = pd.crosstab(
            filtered[approved_col],
            filtered[enrolled_col]
        ).reindex(index=range(10), columns=range(10), fill_value=0)

        sns.heatmap(
            heatmap_data,
            cmap=cmap,
            annot=True,
            fmt='d',
            linewidths=0.5,
            linecolor='#f0f0f0',
            ax=ax,
            cbar_kws={'label': 'Number of Students'}
        )

        ax.set_title(f'{label} — {sem} Sem Enrolled vs Passed', fontsize=13, fontweight='bold', pad=12)
        ax.set_xlabel(f'Units Enrolled ({sem} Sem)', fontsize=11)
        ax.set_ylabel(f'Units Passed ({sem} Sem)', fontsize=11)
        ax.invert_yaxis()

plt.suptitle('Enrolled vs Units Passed — Dropouts vs Graduates',
             fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
# Grafe Distribution of Students Who Dropped Out vs. Graduated

import matplotlib.pyplot as plt
import numpy as np

color_map = {'Dropout': '#E69F00', 'Graduate': '#9B59B6'}
outcomes = ['Dropout', 'Graduate']

# Include ALL students this time, including grade = 0
fig, ax = plt.subplots(figsize=(10, 6))

data = [df[df['Target'] == outcome]['Curricular units 1st sem (grade)'].values for outcome in outcomes]
colors = [color_map[outcome] for outcome in outcomes]

parts = ax.violinplot(data, positions=range(len(outcomes)), showmedians=True, showextrema=True)

for pc, color in zip(parts['bodies'], colors):
    pc.set_facecolor(color)
    pc.set_alpha(0.7)
    pc.set_edgecolor(color)

parts['cmedians'].set_color('white')
parts['cmedians'].set_linewidth(2)
parts['cmaxes'].set_color('grey')
parts['cmins'].set_color('grey')
parts['cbars'].set_color('grey')

# Reference line at 10
ax.axhline(y=10, color='grey', linestyle='--', linewidth=1.5, alpha=0.7, label='Pass threshold (10)')
ax.axhline(y=13, color='black', linestyle='--', linewidth=1.5, alpha=0.7, label='Suggested intervention threshold (13)')

ax.set_xticks(range(len(outcomes)))
ax.set_xticklabels(outcomes, fontsize=12)
ax.set_title('1st Semester Grade Distribution by Student Outcome', fontsize=15, fontweight='bold', pad=14)
ax.set_xlabel('Student Outcome', fontsize=12)
ax.set_ylabel('Average Grade (0–20)', fontsize=12)
ax.spines[['top', 'right']].set_visible(False)
ax.legend(fontsize=10)

plt.tight_layout()
plt.show()
```

First semester grades emerge as one of the clearest early indicators of dropout risk, but the story is more nuanced than a simple pass or fail. The violin plot above reveals two distinct patterns within the dropout population — a significant cluster of students who received a grade of zero, indicating they disengaged before completing a single evaluation, and a broader group whose grades fall below the 13 threshold where graduate students tend to concentrate. The median first semester grade for dropout students sits around 11, compared to approximately 13 for graduates, suggesting that a grade of 13 or above in semester one is a meaningful signal of a student who is on track to complete their program. This points to a clear and actionable intervention window: students finishing semester one below a 13 should be proactively flagged for a check-in with an academic counselor before semester two begins. However, it is equally important to note that the graduate violin also extends down to zero — meaning some students who struggled significantly in semester one went on to graduate anyway. A difficult first semester is not a sentence. With the right support, encouragement, and resources, students can and do recover. Any intervention strategy built around this threshold should lead not with alarm, but with reassurance — letting struggling students know that their first semester does not define their trajectory, and that the institution is invested in helping them find their footing before it is too late.

```python
# Looking for any patterns or changes in students grade from semester 1 to semester 2, comparing dropouts with graduates

import matplotlib.pyplot as plt
import numpy as np

# Filter out students with 0 in both semesters
df_graded = df[
    (df['Curricular units 1st sem (grade)'] > 0) & 
    (df['Curricular units 2nd sem (grade)'] > 0)
].copy()

color_map = {'Dropout': '#E69F00', 'Graduate': '#9B59B6'}

fig, ax = plt.subplots(figsize=(10, 7))

for outcome, color in color_map.items():
    subset = df_graded[df_graded['Target'] == outcome]
    
    for _, row in subset.iterrows():
        ax.plot([0, 1],
                [row['Curricular units 1st sem (grade)'],
                 row['Curricular units 2nd sem (grade)']],
                color=color, alpha=0.03, linewidth=0.8)

    # Bold average line on top
    avg_sem1 = subset['Curricular units 1st sem (grade)'].mean()
    avg_sem2 = subset['Curricular units 2nd sem (grade)'].mean()
    ax.plot([0, 1], [avg_sem1, avg_sem2], color=color, linewidth=3.5,
            label=f'{outcome} (avg: {avg_sem1:.1f} → {avg_sem2:.1f})')
    
    ax.scatter([0, 1], [avg_sem1, avg_sem2], color=color, s=80, zorder=5)

ax.set_xticks([0, 1])
ax.set_xticklabels(['1st Semester', '2nd Semester'], fontsize=12)
ax.set_ylabel('Average Grade (0–20)', fontsize=12)
ax.set_ylim(0, 20)
ax.set_title('Grade Trajectory: 1st vs 2nd Semester by Student Outcome',
             fontsize=15, fontweight='bold', pad=14)
ax.spines[['top', 'right']].set_visible(False)
ax.legend(fontsize=11)

plt.tight_layout()
plt.show()
```

An intuitive assumption when studying student dropout might be that struggling students follow a visible downward trajectory — starting with mediocre first semester grades and then deteriorating further in the second semester as pressures mount. The data does not support this narrative. Among students who were graded in both semesters, dropout students showed virtually no change in average grade between semester one and semester two (12.2 → 12.1), suggesting that academic decline between semesters is not a meaningful driver of dropout. Instead, the more telling signal comes from the students who never received a grade at all — those who disengaged before completing a single evaluation. This reframes the intervention strategy entirely: rather than monitoring for a mid-year grade decline, institutions should focus on identifying students who are already disengaged or underperforming at the end of semester one, because by semester two the pattern is already set.

A great example of the essence of data science: start with intuition and curiosity, then let the data tell the real story and reveal how the narrative unfolds.
```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

log_reg = LogisticRegression(max_iter=1000, random_state=42)
log_reg.fit(X_train_scaled, y_train)

y_pred_log = log_reg.predict(X_test_scaled)
print("Logistic Regression trained.")
```

```python
# Setting out different hyperparameter values to test different values of the max_iter hyperparameter of the ML model, and plotting the results to see how the accuracy changes as the hyperparameter value increases, so that we can find the best value for the hyperparameter
max_iter_values = [10, 25, 50, 100, 200, 500, 1000, 2000, 5000]
results = []

for iters in max_iter_values:
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        model = LogisticRegression(max_iter=iters, random_state=42)
        model.fit(X_train_scaled, y_train)

    results.append({"max_iter": iters, "Accuracy": accuracy_score(y_train, model.predict(X_train_scaled)), "Split": "Train"})
    results.append({"max_iter": iters, "Accuracy": accuracy_score(y_test,  model.predict(X_test_scaled)),  "Split": "Test"})

results_df = pd.DataFrame(results)

fig, ax = plt.subplots(figsize=(10, 6))
sns.lineplot(
    data=results_df, x="max_iter", y="Accuracy", hue="Split",
    marker="o", palette={"Train": "#6c63ff", "Test": "#ff6584"}, ax=ax
)
ax.set_xscale("log")
ax.set_title("Logistic Regression — Accuracy vs max_iter", fontsize=14, fontweight="bold", pad=12)
ax.set_xlabel("max_iter (log scale)", fontsize=11)
ax.set_ylabel("Accuracy", fontsize=11)
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.show()

best_row = results_df[results_df["Split"] == "Test"].sort_values("Accuracy", ascending=False).iloc[0]
```

```python
# Zoomed-in sweep between max_iter=25 and max_iter=200 to pinpoint convergence

zoomed_iter_values = [25, 50, 75, 100, 125, 150, 175, 200]
zoomed_results = []

for iters in zoomed_iter_values:
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        model = LogisticRegression(max_iter=iters, random_state=42)
        model.fit(X_train_scaled, y_train)

    zoomed_results.append({"max_iter": iters, "Accuracy": accuracy_score(y_train, model.predict(X_train_scaled)), "Split": "Train"})
    zoomed_results.append({"max_iter": iters, "Accuracy": accuracy_score(y_test,  model.predict(X_test_scaled)),  "Split": "Test"})

zoomed_df = pd.DataFrame(zoomed_results)

fig, ax = plt.subplots(figsize=(10, 6))
sns.lineplot(
    data=zoomed_df, x="max_iter", y="Accuracy", hue="Split",
    marker="o", palette={"Train": "#6c63ff", "Test": "#ff6584"}, ax=ax
)
ax.set_xticks(zoomed_iter_values)
ax.set_title("Logistic Regression — Accuracy vs max_iter (25–200)", fontsize=14, fontweight="bold", pad=12)
ax.set_xlabel("max_iter", fontsize=11)
ax.set_ylabel("Accuracy", fontsize=11)
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.show()

best_zoomed = zoomed_df[zoomed_df["Split"] == "Test"].sort_values("Accuracy", ascending=False).iloc[0]
print(f"Best max_iter in zoomed range: {int(best_zoomed['max_iter'])}  (accuracy: {best_zoomed['Accuracy']:.4f})")
```

##### Conclusion: 
The zoomed sweep reveals a striking result: the model reaches its peak accuracy of 76.95% at just max_iter=25, and this holds for both the train and test sets simultaneously. Rather than showing a gradual climb that levels off mid-range, accuracy is already at its ceiling within the first 25 iterations and does not improve at all beyond that point. This tells us that the lbfgs solver converges exceptionally quickly on this dataset — even after one-hot encoding expanded the feature space, the optimization problem is well-conditioned enough that the solver finds its best solution in very few steps. The fact that both train and test accuracy peak at the same point also rules out any overfitting dynamic; this is not early stopping providing implicit regularization, the solver simply does not need more iterations to reach its optimum. Based on this sweep, a max_iter of 25 is technically sufficient. In practice, using max_iter=100 is the safer and recommended choice: it guarantees full convergence with a comfortable margin, adds negligible compute time, and avoids convergence warnings that could appear if the data distribution shifts slightly in a different context. The default of 1,000 used in the baseline model above is correct but unnecessarily high — this analysis confirms we could reduce it to 100 and achieve identical results.

#### Step 3: Build a stronger model — Random Forest

Random Forest builds many decision trees and lets them vote on the final prediction. It can capture non-linear patterns and interactions between features (e.g. tuition status mattering differently depending on age or attendance type) better than a linear model. Tree-based models don't need feature scaling, so we use the original (unscaled) `X_train`/`X_test`.

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(n_estimators=200, max_depth=10, random_state=42)
rf.fit(X_train, y_train)

y_pred_rf = rf.predict(X_test)
print("Random Forest trained.")
```

```python
# Sweep max_depth to find where test accuracy peaks before overfitting sets in

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

max_depth_values = [1, 2, 3, 5, 7, 10, 15, 20, 25, 30]
depth_results = []

for depth in max_depth_values:
    model = RandomForestClassifier(n_estimators=200, max_depth=depth, random_state=42)
    model.fit(X_train, y_train)
    depth_results.append({"max_depth": depth, "Accuracy": accuracy_score(y_train, model.predict(X_train)), "Split": "Train"})
    depth_results.append({"max_depth": depth, "Accuracy": accuracy_score(y_test,  model.predict(X_test)),  "Split": "Test"})

depth_df = pd.DataFrame(depth_results)

fig, ax = plt.subplots(figsize=(10, 6))
sns.lineplot(
    data=depth_df, x="max_depth", y="Accuracy", hue="Split",
    marker="o", palette={"Train": "#6c63ff", "Test": "#ff6584"}, ax=ax
)
ax.set_xticks(max_depth_values)
ax.set_title("Random Forest — Accuracy vs max_depth", fontsize=14, fontweight="bold", pad=12)
ax.set_xlabel("max_depth", fontsize=11)
ax.set_ylabel("Accuracy", fontsize=11)
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.show()

best_depth_row = depth_df[depth_df["Split"] == "Test"].sort_values("Accuracy", ascending=False).iloc[0]
best_depth = int(best_depth_row["max_depth"])
print(f"Best max_depth by test accuracy: {best_depth}  (accuracy: {best_depth_row['Accuracy']:.4f})")
```

```python
# Sweep n_estimators using the best max_depth found above

n_estimator_values = [10, 25, 50, 100, 150, 200, 300, 400, 500]
est_results = []

for n_est in n_estimator_values:
    model = RandomForestClassifier(n_estimators=n_est, max_depth=best_depth, random_state=42)
    model.fit(X_train, y_train)
    est_results.append({"n_estimators": n_est, "Accuracy": accuracy_score(y_train, model.predict(X_train)), "Split": "Train"})
    est_results.append({"n_estimators": n_est, "Accuracy": accuracy_score(y_test,  model.predict(X_test)),  "Split": "Test"})

est_df = pd.DataFrame(est_results)

fig, ax = plt.subplots(figsize=(10, 6))
sns.lineplot(
    data=est_df, x="n_estimators", y="Accuracy", hue="Split",
    marker="o", palette={"Train": "#6c63ff", "Test": "#ff6584"}, ax=ax
)
ax.set_xticks(n_estimator_values)
ax.set_title(f"Random Forest — Accuracy vs n_estimators (max_depth={best_depth})", fontsize=14, fontweight="bold", pad=12)
ax.set_xlabel("n_estimators", fontsize=11)
ax.set_ylabel("Accuracy", fontsize=11)
ax.spines[["top", "right"]].set_visible(False)
plt.tight_layout()
plt.show()

best_est_row = est_df[est_df["Split"] == "Test"].sort_values("Accuracy", ascending=False).iloc[0]
print(f"Best n_estimators by test accuracy: {int(best_est_row['n_estimators'])}  (accuracy: {best_est_row['Accuracy']:.4f})")
```

##### Conclusion:
In the graph of accuracy against max_depth, the train accuracy climbs almost vertically — hitting 98% at depth 15 and 100% at depth 20 — while test accuracy rises much more slowly, peaks at max_depth=15 (77.40%), and then actually starts to decline as depth continues to increase. This divergence between the two lines is the signature of overfitting: the trees are growing so deep that they begin memorising training examples rather than learning generalisable patterns. A max_depth of 15 is the sweet spot where the model has learned enough complexity to be useful without memorising noise.

The n_estimators graph, run with max_depth fixed at 15, tells a different and more reassuring story. Train accuracy is essentially flat the entire time, hovering around 98% regardless of how many trees are used. Test accuracy shows a steady, gradual rise from 75.6% at just 10 trees up to a peak of 77.85% at n_estimators=300, after which it plateaus and marginally dips. Adding more trees beyond 300 contributes negligible improvement — the forest has already converged. This is typical behaviour: each additional tree reduces variance a little, but the returns diminish quickly once the ensemble is large enough to average out random fluctuations. Based on this sweep, n_estimators=300 is the recommended setting — it captures the full accuracy benefit without the unnecessary compute cost of 400 or 500 trees.

**Interpretation: comparing the two models** 

The two models reach very similar overall accuracy: Logistic Regression scores 76.9% and Random Forest scores 77.9%. This tells us that a simple linear model already captures most of the useful signal in the data, and that the more complex model adds only a small improvement. That improvement comes mainly from better handling of the two larger classes. Random Forest raises Dropout precision from 0.79 to 0.82 and Dropout F1-score from 0.78 to 0.80, while keeping Graduate recall at a strong 0.93. (Precision means: of all students the model labeled Dropout, how many truly were. Recall means: of all students who truly dropped out, how many the model caught.) The higher Dropout precision is valuable, because Dropout is the class an intervention program would act on, so a more reliable Dropout label is worth more than a small gain elsewhere.
The more important finding is that both models struggle with the middle class, "Enrolled." Logistic Regression correctly identifies only 33% of truly Enrolled students (recall 0.33), and Random Forest only 36%. This means that about two-thirds of still-enrolled students are misclassified as either Dropout or Graduate. The reason is that Enrolled students sit between the two other groups on nearly every predictor we studied: their grades, their approved units, and their financial indicators all overlap heavily with both Dropout and Graduate students. Because of this overlap, the models have very little information that uniquely marks a student as Enrolled. The confusion matrix shows this directly: the Enrolled cases are spread mostly into the Graduate and Dropout predictions rather than forming a clear group of their own. This is the main weakness of the model. It is caused by how similar the classes are in the data, not by poor tuning, which is why neither the hyperparameter sweeps nor the switch to a non-linear model corrected it.

**Final conclusions**

Three main conclusions come out of this project. 

- First, dropout can be detected early. First-semester approved units and grades carry most of the predictive weight, which means a model like this one could be run at the end of the first semester to identify at-risk students while there is still time to help them. This matches the theme from the exploratory analysis: the best time to intervene is early in the program, not at the point of admission. 

- Second, the model fits its intended purpose well, even though the overall accuracy is moderate. Identifying Dropout students and confirming Graduate students is exactly what it does well, with a Dropout F1-score of about 0.80 and a Graduate F1-score of about 0.85. The weak Enrolled performance matters less in practice, because "still enrolled" is a temporary status that is reviewed again the following term. 

- Third, the financial indicators (tuition status, debt, and scholarship) add information that academic features alone do not provide, and they point to specific actions an institution can take rather than outcomes it can only observe.

The limitations should also be stated clearly. We only have grades recorded at the end of each semester, so the model cannot see the decline within a semester that likely comes before a student disengages. Midterm or assignment-level grades would improve both the analysis and the model. The strong class imbalance toward Graduate raises the overall accuracy and is the main cause of the poor Enrolled results, so a fairer evaluation would report the macro-average (the average performance across all three classes, treating each equally) as the primary measure. Future work could also apply class weighting, resampling, or a two-stage model that first separates Dropout from non-Dropout and then divides the rest, in order to improve Enrolled recall. 

Finally, because the data comes from a single institution, the specific thresholds we found, such as the grade-13 line and the 86.6% figure, should be treated as guidance rather than fixed rules until they are tested on other data. These limitations do not weaken the central result: student outcomes here are predictable, can be predicted early, and can be predicted from a small and understandable set of academic and financial measures that an institution could realistically track.
